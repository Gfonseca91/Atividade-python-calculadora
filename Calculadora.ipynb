```python
def somar(num1, num2):
  somar = num1 + num2
  return print(f"A soma dos números{num1} + {num2} é igual a {somar}")

def subtrair(num1, num2):
  somar = num1 - num2
  return print(f"A diferença dos números{num1} - {num2} é igual a {subtrair}")

def dividir(num1, num2):
  if num2 == 0:
    return print("Não é possível realizar essa conta")
  else:
    dividir = num1 / num2
  dividir = num1 / num2
  return print(f"A divisão dos números{num1} / {num2} é igual a {dividir}")

def multiplicar(num1, num2):
  somar = num1 * num2
  return print(f"A multiplicação dos números{num1} * {num2} é igual a {multiplicar}")
```

```python
while True:
  print("\n--- CALCULADORA PYTHON ---")
  print("1 - somar")
  print("2 - subtrair")
  print("3 - dividir")
  print("4 - multiplicar")
  print("5 - SAIR")

  escolha = input(" Escolha uma opção:")

  if escolha == "0":
    print("Calculadore encerrada!")
    break

  if escolha in["1", "2", "3", "4"]:
    num1 = float(input("Digite o primeiro número: "))
    num2 = float(input("Digite o primeiro segundo: "))

    if escolha == "1":
      resultado = somar(num1, num2)

    elif escolha == "2":
      resultado = subtrair(num1, num2)

    elif escolha == "3":
      resultado = dividir(num1, num2)

    elif escolha == "4":
      resultado = multiplicar(num1, num2)
  else:
    print("Escolha um número válido")
```
